### We could also use regex to select words

```python
import re

def select_words(filename:str):
    file = open(filename, 'rt')
    text = file.read()
    file.close
    words = re.split(r'\W+', text)
    return words 

words = select_words(filename="metamorphosis_clean.txt")
first_100_words = words[:100]
print(first_100_words)
```

### Another approach > Use re.punctuation

```python
import string
pattern = re.compile('[%s]' % re.escape(string.punctuation))


stripped_words = [pattern.sub('', word) for word in words]
stripped_words
```

### We can put all of this together, load the text file, split it into words by white space, then translate each word to remove the punctuation.

```python

filename="metamorphosis_clean.txt"

def clean_text(filename):
    file = open(filename, 'rt')
    text = file.read()
    file.close()
    words = text.split()
    re_punc = re.compile('[%s]' % re.escape(string.punctuation))
    re_print = re.compile('[^%s]' % re.escape(string.printable))
    stripped_words = [re_punc.sub(" ", word) for word in words]
    cleaned_words = [re_print.sub(' ', word) for word in stripped_words]
    lower_case_words = [word.lower() for word in cleaned_words]
    return lower_case_words

clean_text(filename)
```
